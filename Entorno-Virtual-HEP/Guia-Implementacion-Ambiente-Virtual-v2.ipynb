## 8.1 Registrar el ambiente como kernel de Jupyter

Después de instalar Jupyter, ejecuta:

```bash
python -m ipykernel install --user \
    --name fisica-computacional \
    --display-name "Python 3.10 (Física Computacional)"
```

Esta instrucción registra el ambiente `fisica-computacional` para que pueda seleccionarse como **kernel** dentro de Jupyter y Visual Studio Code.

Las partes principales son:

```text
python -m ipykernel install
```

ejecuta el módulo `ipykernel` y registra un nuevo kernel;

```text
--user
```

realiza el registro para el usuario actual;

```text
--name fisica-computacional
```

establece el nombre interno del kernel;

```text
--display-name "Python 3.10 (Física Computacional)"
```

define el nombre que veremos en la interfaz de Jupyter o VS Code.

En conjunto:

```text
ambiente fisica-computacional
          ↓
registrar como kernel
          ↓
Python 3.10 (Física Computacional)
          ↓
seleccionable en Jupyter y VS Code
```

### Windows 11

En PowerShell, escribe el mismo comando en una sola línea:

```powershell

```

> **Nota:** instalar Jupyter proporciona las herramientas para trabajar con notebooks; registrar `ipykernel` permite que esos notebooks utilicen específicamente el Python del ambiente `fisica-computacional`.

## 17.1 Particle

Brinda información referente a las partículas, esto mediante el Particle Data Group, el cual define la identificación estandar de partículas (ID) [1]. Entonces, mediante su ID podemos acceder a su información, tal como: 

- Masa.

- Vida media (lifetime), cuando está disponible.

- Carga eléctrica.

- Espín.

- Paridad (P).

- Paridad de conjugación de carga (C).

Entre otras propiedades.

````python
#Importamos la librería
from particle import Particle

#Definimos los objeotos a partir de su ID
muon = Particle.from_pdgid(13)
pion = Particle.from_pdgid(211)
proton = Particle.from_pdgid(2212)

#Imprimimos la información requerida
print(muon)
print("Masa del muón [MeV]:", muon.mass)
print("Carga del muón:", muon.charge)

print(pion)
print(proton)
````

## 17.2 Vector

Vector es una biblioteca de Python para vectores espaciales en 2D y 3D, así como para vectores espacio-temporales en 4D. Está diseñada especialmente para realizar cálculos geométricos en matrices de vectores, en lugar de procesar un vector a la vez en un bucle «for» de Python [2]. Además, su enfoque es hacia ffísica de altas energías (HEP), por lo que su manejo en esta área es natural, permitiendo realizar cálculos relativistas.

````python
#Importamos la librería
import vector

#Definimos los cuadrivectores asociados a muones
mu1 = vector.obj(px=40, py=0, pz=30, E=60)
mu2 = vector.obj(px=-40, py=0, pz=-30, E=60)

#Definimos el cuadrivector del sistema
sistema = mu1 + mu2

#Imprimimos la información requerida
print("Masa invariante [GeV]:", sistema.mass)                   #Se observa que podemos acceder a la masa total del sistema
print("Energía [GeV]:", sistema.E)                              #Se observa que podemos acceder a la energía total del sistema
````

## 17.3 hepunits

hepunits de Scikit-HEP es un paquete de Python que proporciona unidades físicas utilizadas habitualmente en física de altas energías.

Su objetivo es facilitar el manejo de magnitudes físicas y evitar errores al trabajar con diferentes escalas de unidades:

- Energía: eV, keV, MeV, GeV, TeV.
- Longitud: mm, cm, m, km.
- Tiempo: ns, μs, ms, s.
- Masa: kg y unidades equivalentes de energía en física de partículas.
- Ángulos: radianes y grados.
- Área: barn, millibarn, microbarn, nanobarn y picobarn.

````python
#Importamos la biblioteca 
import hepunits as u

#Definimos las cantidades con sus unidades
energia = 5 * u.GeV
masa = 500 * u.MeV

#Imprimimos la información requerida
print("Energía:", energia)
print("Masa:", masa)
print("Masa en GeV:", masa / u.GeV)         #Notamoos que la operación es consistente con las unidades
````

## 17.4 pyhepmc

El paquete pyhepmc es la versión de python de HEPMC. El paquete HepMC es un registro de eventos orientado a objetos, escrito en C++, para generadores Monte Carlo y simulaciones en el ámbito de la física de altas energías. Por lo tanto, pyhepmc permite leer, crear, modificar y escribir eventos de física de altas energías utilizando el formato y las estructuras de HepMC3. [4] De esta manera, podemos definir un evento y acceder a su información:

- Eventos: Cada colisión o interacción simulada.
- Partículas: Partículas iniciales y finales de cada evento.
- Cuadrimomentos: Energía y componentes del momento.
- Vértices: Puntos espaciales y temporales donde ocurren interacciones o decaimientos.
- Relaciones entre partículas: Qué partículas producen qué otras en un decaimiento.
- Información del evento: Número de eventos, pesos y atributos adicionales.
- Unidades: Unidades de energía y longitud usadas en el archivo.

````python
import pyhepmc

#Instanciamos un evento y un vértice
evento = pyhepmc.GenEvent()
vertice = pyhepmc.GenVertex()

# Instanciamos una partícula: Bosón de Higgs
higgs = pyhepmc.GenParticle(
    pyhepmc.FourVector(0, 0, 0, 125),
    25,
    2
)

# Instanciamos una partícula Gamma
gamma1 = pyhepmc.GenParticle(
    pyhepmc.FourVector(0, 0, 62.5, 62.5),
    22,
    1
)

#Instanciamos otra partícula Gamma
gamma2 = pyhepmc.GenParticle(
    pyhepmc.FourVector(0, 0, -62.5, 62.5),
    22,
    1
)

#Añadimos información al vértice: partículas entrantes y salientees
vertice.add_particle_in(higgs)
vertice.add_particle_out(gamma1)
vertice.add_particle_out(gamma2)

#Añadimos el vértice al evento
evento.add_vertex(vertice)

#Definido el evento, podemos guardarlo
with pyhepmc.open("evento.hepmc", "w") as archivo:
    archivo.write(evento)

print("Evento guardado")
````

````python
#Leemos el evento guardado anteriormente
with pyhepmc.open("evento.hepmc") as archivo:
    for evento in archivo:
        #imprimimos la información del evento
        print(evento)
````

## 17.6 pylhe
Interfaz de python que permite leer archivos 'Les Houches Event' (LHE), el cual es un formato para representar un evento. Así, nos recuerda a HepMC3, el cual contiene información más explicita que LHE, sin embargo LHE permite el intercambio de información entre diferentes programas de simulación o cálculo [6]. Así como las imagenes tienen diferentes formatos, los eventos también tienen diferentes formatos y es importante destacar su utilidad.

````python
#Importamos las librerias
from skhep_testdata import data_path
import pylhe

#Extraemos la ruta a un archivo LHE proveniente de skhep_testdata
ruta = data_path("pylhe-testlhef3.lhe")

#Leemos el archivo LHE proveniente de la ruta
archivo = pylhe.LHEFile.fromfile(ruta)

#Extraemos el número de eventos que contiene dicho archivo
print("Número de eventos:", pylhe.LHEFile.count_events(ruta))

#evento almacena el primer evento del archivo: mediante 'next()'
evento = next(archivo.events)

#particula recorre las particulas involucradas en el primer evento.
for particula in evento.particles:
    #Imprimimmos la información referente a las partículas del primer evento
    print(
        "PDG:", particula.id,
        "px:", particula.px,
        "py:", particula.py,
        "pz:", particula.pz,
        "E:", particula.e
    )
````

## 17.7 phasespace

El paquete phasespace permite generar eventos de desintegración sin necesidad de instalar ROOT, mediante una implementación basada en TensorFlow. Además, incorpora la generación de cadenas de desintegración complejas y el tratamiento de resonancias, cuyas masas no son fijas, sino que siguen una distribución [7]. De esta manera, podemos generar un proceso de decaimiento con momentos finales aleatorios pero pertenecientes a un espacio de fase posible. 

````python
#Importamos la biblioteca
import phasespace

# Definimos las partículas con sus masas en GeV
B0 = phasespace.GenParticle("B0", 5.279)

K = phasespace.GenParticle("K+", 0.494)
pi1 = phasespace.GenParticle("pi-", 0.140)
pi2 = phasespace.GenParticle("pi0", 0.135)

# Construimos el proceso de decaimineto: B0 -> K+ pi- pi0
B0.set_children(K, pi1, pi2)

# Generamos 1000 eventos
pesos, particulas = B0.generate(n_events=1000)

#Imprimimos el número de eventos generados y la partículas involucradas
print("Eventos generados:", len(pesos))
print("Partículas:", particulas.keys())

#Recorre las partículas involucradas, extrayendo extrae sus nombres y los cuadrimomentos de los 1000 eventos
for nombre, p4 in particulas.items():
    #Imprimimos el nombre de la partícula
    print(f"\n{nombre}")
    #Imprimos la forma del arreglo que almacena los cuadrimomentos
    print("Forma:", p4.shape)
    #Imprimos el momento del primer evento
    print("Primer evento:", p4[0])
````

## 17.8 fastjet

Para entender este paquete, es necesario entender a que hace referencia: jet. Los jet son los chorros colimados de hadrones que resultan de la fragmentación de un quark o un gluón de alta energía. Suelen ser estructuras visualmente evidentes cuando se observa la visualización de un evento experimental, y al medir su energía y dirección se puede aproximar la idea de su origen [8].

Luego, lo que permite fastjet es reconstruir jets a partir de las partículas producidas en colisiones de alta energía. Para ello, implementa algoritmos de agrupamiento (jet clustering), como ($k_t$), anti-($k_t$) y Cambridge/Aachen. Para fines ilustrativos podemos explicar el algoritmo anti-($k_t$) y mostrar su implementación.

### 17.8.1 Algoritmo anti-($k_t$)
Dadas las partículas finales A, B, C. Podemos definir la 'distancia' entre partículas:

\begin{align}
    d_{ij} = \min{\bigg(\frac{1}{p^2_{ti}}, \frac{1}{p^2_{ti}}\bigg)} \frac{\Delta R^2_{ij}}{R^2}
\end{align}

donde:
- $p_{ti}$: Momento transversal de la partícula 'i', el momento transversal $p_t$ es la componente del momento de una partícula perpendicular a la dirección de los haces de partículas. En el LHC, las colisiones suceden a lo largo del eje Z, por lo tanto este eje es paralelo al haz: $p_{ti} = \sqrt{p^2_x + p^2_y} $.
- R: parámetro que controla el tamaño característico del jet. Usualmente igual a 1.
- $\Delta R_{ij}$: separación entre las partículas en el espacio de rapidez y ángulo azimutal.

Al calcular d_{ij}, podemos realizarlo entre partículas (partícula - partículas) y con respecto al haz (partículas - haz). De este conjunto, pueden pasar dos cosas:

- La menor distancia es entre partículas: Supongamos A y B, entonces se define un pseudo-jet AB. Nuevo conjunto: AB, C
- La menor distancia es entre partícula y haz: Supongamos A, entonces se define el jet A y se deja fuera del algoritmo. Nuevo conjunto: B, C

En ambos casos, se vuelve a determinar las distancias $d_{ij}$ hasta que solo se tenga jets.

````python
#Importamos el paquete
import fastjet

#Definimos las partículas finales
particulas = [
    fastjet.PseudoJet(30, 0, 10, 35),
    fastjet.PseudoJet(25, 5, 8, 28),
    fastjet.PseudoJet(-20, 0, 0, 20),
]

#Definimos el algoritmo utilizado como anti-kt
definicion = fastjet.JetDefinition(
    fastjet.antikt_algorithm,
    0.4
)

#Construimos la secuencia de agrupamiento
secuencia = fastjet.ClusterSequence(particulas, definicion)
#Obtenemos los jet finales y los ordenamos por momento transversal
jets = fastjet.sorted_by_pt(secuencia.inclusive_jets())

#Imprimimos los jet con sus momentos transversales y angulos polares medidos desde el eje Z.
for i, jet in enumerate(jets):
    print(f"Jet {i}: pT={jet.pt():.2f} GeV, eta={jet.eta():.2f}")
````

## 17.9 awkward

Es una biblioteca de Python diseñada para trabajar con datos estructurados de longitud variable, muy comunes en Física de Altas Energías.

````python
#Importamos el paquete 
import awkward as ak

#Definimos un awkward array, tal que represente el momento en el eje 'x' de cada partícula que conforme cada instante: 2 partículas -> 3 partículas -> 1 partícula 
px = ak.Array([
    [10, 20],
    [10, 8, 12],
    [30]
])

#Imprimimos el número de partículas que conforman cada evento
print("Número de partículas por evento:", ak.num(px))
#Imprimimos el primer evento
print("Primer evento:", px[0])

# Suma de px de las partículas de cada evento
print("Suma px:", ak.sum(px, axis=1))
````

## 17.11 hist

Hist es una biblioteca de Python del ecosistema Scikit-HEP diseñada para crear, manipular y analizar histogramas, especialmente en análisis de física de altas energías (HEP). [10]

````python
#Importamos los paquetes
import hist
import numpy as np
import matplotlib.pyplot as plt

#Creamos el histograma tal que hace la referencia a la masa en GeV: 50 bins regulares entre 0 y 100
h = hist.Hist(
    hist.axis.Regular(50, 0, 100, name="masa", label="Masa [GeV]")
)

#Generamos 10000 datos con desviación estándar igual a 10 y media 50
datos = np.random.normal(50, 10, 10000)
#Guardamos los datos en el histograma
h.fill(datos)

#Imprimimos el total de entradas y los conteos
print("Total de entradas:", h.sum())
print("Conteos:", h.values())

#Mostramos el histograma
h.plot()
plt.show()
````

## 17.12 qrules

Es un paquete de Python para validar y generar reacciones de partículas utilizando las reglas de conservación de los números cuánticos. El usuario solo tiene que proporcionar un conjunto determinado de condiciones de contorno (estado inicial y final, tipos de interacción permitidos, topologías de desintegración esperadas, etc.). A continuación, QRules abarcará el espacio de los números cuánticos permitidos en todas las topologías de desintegración permitidas y las instancias de partículas que correspondan con los conjuntos de números cuánticos permitidos que haya encontrado. [11]

````python
# Importamos el paquete
import qrules

# Definimos y generamos las transiciones
transiciones = qrules.generate_transitions(
    initial_state=["J/psi(1S)"],                     #Estado inicial
    final_state=["gamma", "pi0", "pi0"],             #Estado final
    allowed_intermediate_particles=[                 #Especifica las partículas intermedias permitidas
        "f(0)(980)",
        "f(2)(1270)"
    ],
    formalism="helicity"                             #Especificamos el formalismo para describir las transiciones
)

# Mostrar resultados
print("=" * 50)
print("ANÁLISIS DE TRANSICIONES")
print("=" * 50)

# Indicamos el número total de transiciones posibles
print(f"\nTransiciones encontradas: {len(transiciones.transitions)}")

# Indicamos las partículas intermedias
print("\nPartículas intermedias:")
print(transiciones.get_intermediate_particles().names)

# Mostramos cada transición
for i, transicion in enumerate(transiciones.transitions, start=1):
    print(f"\nTransición {i}:")
    print(transicion)

print("\n" + "=" * 50)
````

## 17.13 ampform

Es un pquete que nos permite partir de las transiciones permitidas y obtener la amplitud del decaimiento, particularmente del análisis de ondas parciales [12]. Su uso no es complicado debido a que sirve de complemento de qrules.

````python
#Importamos los paquetes
import qrules
import ampform

# Definimos y generamos las transiciones
transiciones = qrules.generate_transitions(
    initial_state=["J/psi(1S)"],
    final_state=["gamma", "pi0", "pi0"],
    allowed_intermediate_particles=["f(0)(980)"],
    formalism="helicity"
)

# Indicamos el número total de transiciones posibles
print("Transiciones encontradas:", len(transiciones.transitions))

# Mostramos cada transición
for i, transicion in enumerate(transiciones.transitions, start=1):
    print(f"\nTransición {i}:")
    print(transicion)

# Construimos el modelo de amplitud
modelo = ampform.get_builder(transiciones).formulate()

print("\nModelo de amplitud:")
print(modelo)
````

## 17.14 tensorwaves

Es una biblioteca de Python que permite evaluar numéricamente modelos de amplitudes y realizar análisis de amplitudes en física de partículas [13].

Entonces, el flujo de trabajo habitual sería:

1. qrules: Obtenemos las posibles transiciones
2. ampform: Obtenemos la amplitud de las transiciones.
3. tensorwaves: evaluamos numéricamente la amplitud.

````python
# Importamos los paquetes
import tensorwaves
from tensorwaves.function.sympy import create_parametrized_function

# Desarrollamos la expresión simbólica de la intensidad
intensidad = modelo.intensity.doit()

# Convertimos la expresión a una función numérica
funcion = create_parametrized_function(
    intensidad,
    parameters=modelo.parameter_defaults,
    backend="numpy"
)

print("\nFunción numérica:")
print(funcion)
````

```text
pip          → paquetes instalados mediante pip
```

El archivo funciona como una **descripción reproducible del entorno**.
