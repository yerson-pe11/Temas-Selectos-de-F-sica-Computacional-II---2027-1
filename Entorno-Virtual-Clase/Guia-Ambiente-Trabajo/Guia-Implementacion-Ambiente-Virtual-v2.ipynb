## Propósito

Esta guía establece un entorno de trabajo uniforme para el curso, de modo que todos los estudiantes dispongan de una configuración reproducible y adecuada para programación científica, análisis de datos y física de partículas.

La arquitectura general será:

```text
Sistema operativo
        │
        ▼
     Miniconda
        │
        ▼
Ambiente virtual Conda
        │
        ├── Python 3.10
        ├── NumPy
        ├── SciPy
        ├── Matplotlib
        ├── Pandas
        ├── SymPy
        ├── Scikit-learn
        ├── Jupyter
        │
        └── Scikit-HEP
             ├── uproot
             ├── awkward
             ├── vector
             ├── hist
             ├── mplhep
             └── particle
        │
        ▼
Visual Studio Code
```

> **Importante**  
> No es necesario instalar Python desde `python.org`.  
> La versión **Python 3.10** se instalará y administrará dentro del ambiente de Conda.

# 2. Verificar la instalación

Antes de continuar, asegúrate de que el ambiente:

```text
(fisica-computacional)
```

esté activo.

Las siguientes instrucciones comprueban que Python puede **importar correctamente las bibliotecas instaladas**.

La estructura general:

```bash
python -c "..."
```

indica a Python que ejecute directamente el código escrito entre comillas, sin necesidad de crear un archivo `.py`.

---

## 2.1 Bibliotecas científicas

Ejecuta:

```bash
python -c "import numpy, scipy, matplotlib, pandas, sympy, sklearn; print('Stack científico: OK')"
```

La instrucción:

```text
import numpy, scipy, matplotlib, pandas, sympy, sklearn
```

intenta cargar las principales bibliotecas científicas que utilizaremos en el curso.

Si todas pueden importarse correctamente, Python ejecutará:

```text
print('Stack científico: OK')
```

y obtendrás:

```text
Stack científico: OK
```

El flujo de la comprobación es:

```text
Python
  ↓
importar bibliotecas
  ↓
¿se cargaron correctamente?
  ↓
Stack científico: OK
```

Si aparece un error como:

```text
ModuleNotFoundError
```

significa que alguna biblioteca no está disponible en el Python que se está utilizando.

## 9. Prueba con partículas

Crea `prueba_particula.py`:

```python
from particle import Particle

electron = Particle.from_pdgid(11)
muon = Particle.from_pdgid(13)
top = Particle.from_pdgid(6)

print(electron)
print(muon)
print(top)
```

`Particle.from_pdgid(...)` identifica una partícula mediante su **PDG ID**:

```text
11 → electrón
13 → muón
 6 → quark top
```

Ejecuta:

```bash
python prueba_particula.py
```


````python
from particle import Particle

electron = Particle.from_pdgid(11)
muon = Particle.from_pdgid(13)
top = Particle.from_pdgid(6)

print(electron)
print(muon)
print(top)
````

## 10. Prueba con cuadrivectores

Crea `prueba_vector.py`:

```python
import vector

p = vector.obj(
    px=30.0,
    py=40.0,
    pz=100.0,
    E=120.0,
)

print("pT =", p.pt)
print("|p| =", p.p)
print("masa invariante =", p.mass)
```

`vector.obj(...)` construye un cuadrivector a partir de las componentes del momento y la energía.

```text
p.pt   → momento transversal
p.p    → magnitud del trimomento
p.mass → masa invariante
```

Ejecuta:

```bash
python prueba_vector.py
```


````python
import vector

p = vector.obj(
    px=30.0,
    py=40.0,
    pz=100.0,
    E=120.0,
)

print("pT =", p.pt)
print("|p| =", p.p)
print("masa invariante =", p.mass)
````
